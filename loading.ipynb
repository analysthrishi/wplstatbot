```python
import json
import pandas as pd

wicket_rows = []

for _, row in df_balls.dropna(subset=['wickets']).iterrows():

    # Convert JSON string → Python list
    wicket_events = json.loads(row['wickets'])

    for w in wicket_events:
        batter_out = w.get("player_out")
        kind = w.get("kind")

        # Some dismissals have no fielders
        fielders = w.get("fielders", [])

        # If no fielders, still create ONE row
        if not fielders:
            wicket_rows.append({
                "ball_id": row["ball_id"],
                "batter_out": batter_out,
                "kind": kind,
                "bowler": row["bowler"],
                "fielder": None,
                "team": row["team"],
                "over_number": row["over_number"]
                "ball": row["over_number"]
            })
        else:
            # One row per fielder (important for catches)
            for f in fielders:
                wicket_rows.append({
                    "ball_id": row["ball_id"],
                    "batter_out": batter_out,
                    "kind": kind,
                    "bowler": row["bowler"],
                    "fielder": f.get("name"),
                    "team": row["team"],
                    "over_number": row["over_number"]
                })
```

```python
df_wickets = pd.DataFrame(wicket_rows)
df_wickets.head()
```

```python
df_balls = []

for match_id, match in enumerate(all_data, start=1):

    balls = pd.json_normalize(
        match,
        record_path=['innings', 'overs', 'deliveries'],
        meta=[
            ['innings', 'team'],
            ['innings', 'overs', 'over']
        ]
    )

    balls = balls.rename(columns={
        'innings.team': 'batting_team',
        'innings.overs.over': 'over_number'
    })

    balls['match_id'] = match_id
    df_balls.append(balls)

df_balls = pd.concat(df_balls, ignore_index=True)

df_balls.insert(0, 'ball_id', range(1, len(df_balls) + 1))
df_balls['over_number'] = df_balls['over_number'].astype(float)
df_balls.info()
```

```python
import sqlite3

conn = sqlite3.connect('wpl.db')
cursor = conn.cursor()

# Get names of all tables
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()

for table in tables:
    table_name = table[0]
    print(f"\nTable: {table_name}")
    
    # PRAGMA table_info returns (id, name, type, notnull, default_value, pk)
    cursor.execute(f"PRAGMA table_info({table_name})")
    for col in cursor.fetchall():
        print(f"  - {col[1]} ({col[2]})")

conn.close()
```
